## 3. ***Data Wrangling***

### Data Wrangling Code

डेटा व्रैंगलिंग (Data Wrangling) के मुख्य चरणों का संक्षेप में विवरण:

1.डेटा कलेक्शन (Data Collection):
विभिन्न स्रोतों (CSV, डेटाबेस, APIs) से डेटा इकट्ठा करें।

2.डेटा अंडरस्टैंडिंग (Data Understanding):
डेटा का प्रारूप, वेरिएबल्स, और संरचना को समझें।
डेटा का निरीक्षण करें (head(), describe(), info())।

3.डेटा क्लीनिंग (Data Cleaning):
Missing Values: Null/NaN मानों को संभालें (ड्रॉप या इम्प्यूट)।
Duplicates: डुप्लिकेट रिकॉर्ड्स हटाएं।
Outliers: असामान्य मानों का पता लगाएं और सही करें।
Inconsistent Formats: यूनिट्स, डेट फॉर्मेट, या स्ट्रिंग फॉर्मेट को स्टैंडर्ड बनाएं।

4.डेटा ट्रांसफॉर्मेशन (Data Transformation):
Scaling/Normalization: संख्यात्मक डेटा को सामान्य सीमा में लाएं।
Encoding: कैटेगोरिकल वेरिएबल्स को एनकोड करें (Label/One-Hot Encoding)।
Feature Engineering: नए वेरिएबल्स बनाएं।

5.डेटा इंटीग्रेशन (Data Integration):
विभिन्न डेटा स्रोतों को मर्ज करें।
रिलेशनशिप बनाएं (join, merge)।
डेटा रिडक्शन (Data Reduction):

अनावश्यक वेरिएबल्स या रिकॉर्ड्स हटाएं।
PCA, Feature Selection का उपयोग करें।
डेटा वेलिडेशन (Data Validation):

डेटा की क्वालिटी और कंसिस्टेंसी सुनिश्चित करें।
लॉजिक और बिजनेस रूल्स की जांच करें।
डेटा सेविंग (Data Saving):

तैयार डेटा को फाइल (CSV, JSON) या डेटाबेस में सेव करें।
इन चरणों का उद्देश्य डेटा को एनालिसिस के लिए स्वच्छ और संरचित बनाना है।
